```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
df=pd.read_csv('D:/Data/pizza_sales.csv')
df
```

```python
df['Year']=df['order_date'].dt.year
```

```python
df['Month_Name']=df['order_date'].dt.strftime('%b')
```

```python
plt.figure(figsize=(10,5))
sns.lineplot(y='total_price',x='Month_Name',data=df,color='g')
plt.title('Monthly Trend')
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.boxplot(y='total_price',x='pizza_category',data=df,gap=2, whis=1.5)
plt.title('Pizza by Category')
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.displot(df['quantity'],bins=[1.0,1.5,2.0,2.5,3.0,3.5,4.0]cc)
plt.title('per unit quantity')
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.countplot(data=df,x='pizza_category',order=df['pizza_category'].value_counts().index)
plt.title('Number of order by pizza category')
plt.xlabel('Pizza Category')
plt.ylabel('Order Count')
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.barplot(data=df,y='total_price',x='pizza_size',order=['S','M','L','XL','XXL'])
plt.title('Average total price by pizza size')
plt.xlabel('Pizza Size')
plt.ylabel('Average Total Price')
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.boxplot(data=df,y='unit_price',x='pizza_category')
plt.title('unit price distribution by category')
plt.show()
```

```python
monthly_sales=df.groupby('Month_Name',as_index = False)['total_price'].sum()
# fix month order
month_order=['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
monthly_sales['Month_Name']=pd.Categorical(monthly_sales['Month_Name'],categories=month_order,ordered=True)
monthly_sales=monthly_sales.sort_values('Month_Name')

plt.figure(figsize=(10,5))
sns.lineplot(data=monthly_sales,x='Month_Name',y='total_price',marker='o')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.histplot(data=df,x='quantity',bins=10,kde=True)
plt.title('Distribution of Quantity per Order')
plt.xlabel('Quantity')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.countplot(data=df,x='pizza_size',order=df['pizza_size'].value_counts().index)
plt.title('Order Count by Pizza size')
plt.xlabel('pizza size')
plt.ylabel('count')
plt.show()
```

```python
top5=df.groupby(['pizza_name'],as_index=False)['quantity'].sum().sort_values('quantity',ascending=False).head(5)
plt.figure(figsize=(10,5))
sns.barplot(data=top5,y='pizza_name',x='quantity')
plt.title('Top 5 Best-Selling Pizza by Quantity')
plt.xlabel('Total Quantity Sold')
plt.ylabel('Pizza name')
plt.show()
```

```python
pivot=df.pivot_table(values='total_price',index='pizza_category',columns='pizza_size',aggfunc='mean')
pivot=pivot[['S','M','L','XL','XXL']]

plt.figure(figsize=(8,5))
sns.heatmap(pivot,annot=True,fmt='.1f',cmap='viridis',mask=pivot.isnull(),linewidths=0.5,linecolor='white')
plt.title('Average Total Price: Category vs Size')
plt.xlabel('Pizza Size')
plt.ylabel('Pizza Category')
plt.figtext(0.5,-0.02,'Note: XL/XXL only offered in the Classic category',ha='center',fontsize=9,style='italic')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.violinplot(y='total_price',x='Month_Name',data=df,palette='plasma')
plt.title('Monthly Trend')
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.violinplot(data=df,y='total_price',x='pizza_size',order=['S','M','L','XL','XXL'],palette='plasma')
plt.title('Average total price by pizza size')
plt.xlabel('Pizza Size')
plt.ylabel('Average Total Price')
plt.show()
```
